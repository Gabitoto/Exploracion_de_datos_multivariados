```r

library(readxl)

df <- read_excel("Secundario_diamante_adultos.xlsx")
table(df$division)
nrow(df)
table(df$NotaDiciembre)

library(dplyr)


```

Se borra la columna curso y se cambia el nombre a la de division por curso y a la de suborganizacion por institucion


```r
library(dplyr)
df <- df %>% select(-curso)
df <- df %>% rename( institucion = suborganizacion)
df <- df %>% rename( curso = division)
```

```r
table(df$Asignatura)
```

```r
library(dplyr)


df <- df %>%  filter(Asignatura %in% c("GEOGRAFIA", "HISTORIA", "MATEMATICA", "LENGUA Y LITERATURA", "LENGUAS EXTRANJERAS" ))
table(df$Asignatura)

```

```r
library(dplyr)

table(df$NotaDiciembre)
table(df$NotaPrimerTrimestre)
table(df$NotaSegundoTrimestre)
table(df$NotaTercerTrimestre)
table(df$NotaDiciembre)

df[df == "SIN NOTA"] <- NA


table(df$NotaPrimerTrimestre)
table(df$NotaSegundoTrimestre)
table(df$NotaTercerTrimestre)
table(df$NotaDiciembre)




# Contar el número de NAs por columna
na_por_columna <- colSums(is.na(df))

# Mostrar el conteo de NAs por columna
print(na_por_columna)

table(df$NotaDiciembre)

```

```r

library(dplyr)


# Filtrar las filas que cumplen la condición
df_limpio <- df %>%
  filter(!if_any(c(NotaPrimerTrimestre, NotaSegundoTrimestre, NotaTercerTrimestre), is.na) |
           (is.na(NotaDiciembre) & !is.na(NotaPrimerTrimestre) & !is.na(NotaSegundoTrimestre) & !is.na(NotaTercerTrimestre)))



# Contar el número de NAs por fila
na_por_columna <- colSums(is.na(df_limpio))

# Mostrar el conteo de NAs por fila
print(na_por_columna)

```

```r
df_limpio <- df_limpio %>%
  mutate(NotaPrimerTrimestre = as.integer(NotaPrimerTrimestre),
         NotaSegundoTrimestre = as.integer(NotaSegundoTrimestre),
         NotaTercerTrimestre = as.integer(NotaTercerTrimestre),
         NotaDiciembre = as.integer(NotaDiciembre))
summary(df_limpio)


```

```r
library(dplyr)

df_limpio <- df_limpio %>%
  mutate(Promedio = (NotaPrimerTrimestre + NotaSegundoTrimestre + NotaTercerTrimestre) / 3,
         Aprobado = ifelse(Promedio < 6, "Reprobo promedio",
                 ifelse(Promedio >= 6 & NotaTercerTrimestre < 6, "Reprobo tercer trimestre",
                        "Aprobado")),
         ResultadoDiciembre = case_when(
           Promedio >= 6 & NotaTercerTrimestre >= 6 ~ "Aprobado en cursada",
           (Promedio >= 6 & NotaTercerTrimestre < 6 & NotaDiciembre >= 6) | (Aprobado == "Reprobo promedio" & NotaDiciembre >= 6) ~ "Aprobado en mesa",
           (Promedio >= 6 & NotaTercerTrimestre < 6 & (is.na(NotaDiciembre) | NotaDiciembre == 1)) ~ "Reprobo(no se presento a mesa)",
           (Promedio < 6 & (is.na(NotaDiciembre) | NotaDiciembre == 1)) ~ "Reprobo(no se presento a mesa)", # Modified this condition to include NotaDiciembre == 1
           Aprobado == "Reprobo promedio" & NotaDiciembre <= 5 ~ "Reprobo en mesa",
           Aprobado == "Reprobo tercer trimestre" & NotaDiciembre <= 5 ~ "Reprobo en mesa",
           TRUE ~ as.character(NA) # Added a default case to handle any unclassified rows
         ))

no_aprobados <- df_limpio %>%
  filter(Aprobado == "Reprobo promedio" | Aprobado == "Reprobo tercer trimestre") %>%
  nrow()

print(no_aprobados)

table(df_limpio$ResultadoDiciembre)
df_limpio
library(dplyr)
no_rinde <- df_limpio %>% filter(ResultadoDiciembre == "No Rinde")
```

```r
library(dplyr)


nrow(df_limpio)
duplicados <- df_limpio %>%
  group_by(across(everything())) %>%
    filter(n() > 1) %>%
    ungroup()
df_sin_duplicados <- anti_join(df, duplicados, by = "Documento")



nrow(df_limpio)
```

```r
library(dplyr)


nrow(df_limpio)
# Filtrar los documentos que aparecen menos de 6 veces o mas de 8
documentos_a_eliminar <- df_limpio %>%
  group_by(Documento) %>%
  filter(n() != 5) %>%
  pull(Documento) %>%
  unique()

# Eliminar los registros con esos documentos
df_1_filtrado <- df_limpio %>%
  filter(!(Documento %in% documentos_a_eliminar))

nrow(df_1_filtrado)

```

```r
library(dplyr)

# Agrupar por 'Documento' y contar registros
alumnos_distintos_5 <- df_1_filtrado %>%
  group_by(Documento) %>%
  summarise(Cantidad_registros = n()) %>%
  filter(Cantidad_registros != 5)

# Mostrar la lista de alumnos y la cantidad de registros
print(alumnos_distintos_5)
```
