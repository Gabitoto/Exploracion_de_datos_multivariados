```r
# prompt: cargar el archivo Primario_diamante.xlsx, sin lineas comentadas

library(readxl)
df <- read_excel("Primario_diamante.xlsx")
head(df)
```

Reviso datos faltantes


```r
# prompt: quiero revisar si hay datos faltantes en la tabla, quiero que me aparezca cada columna y la cantidad de datos faltantes contados

library(dplyr)
df %>%
  summarise_all(~sum(is.na(.)))
```

```r
# prompt: eliminar los datos faltantes

df <- df %>% na.omit()

# Reviso datos faltantes de nuevo
df %>%
  summarise_all(~sum(is.na(.)))
```

Reviso la cantidad de casos de epac para tomar una desicion

```r
library(dplyr)



df <- df %>% select(-curso)
df <- df %>% rename( institucion = suborganizacion)
df <- df %>% rename( curso = division)

summary(df)
```

```r
table(df$NotaPrimerTrimestre)
table(df$NotaSegundoTrimestre)
table(df$NotaTercerTrimestre)
table(df$NotaDiciembre)
```

Al no haber registros con uno que puedan conciderarse inasistencias, decido transformar los EPAC en 1 y los SIN NOTA y los PA seran borrados al ser infima la cantidad y por ultimo elimino la columna Nota diciembre, ya que no esta completa con SIN NOTA




```r
# prompt: quiero transformar los EPAC en 1 y Eliminar los PA y los SIN NOTA, en el orden que convenga y elimino la columna NotaDiciembre

df <- df %>%
  mutate(NotaPrimerTrimestre = ifelse(NotaPrimerTrimestre == "EPAC", 1, NotaPrimerTrimestre),
         NotaSegundoTrimestre = ifelse(NotaSegundoTrimestre == "EPAC", 1, NotaSegundoTrimestre),
         NotaTercerTrimestre = ifelse(NotaTercerTrimestre == "EPAC", 1, NotaTercerTrimestre)) %>%
  filter(NotaPrimerTrimestre != "SIN NOTA" & NotaPrimerTrimestre != "PA",
         NotaSegundoTrimestre != "SIN NOTA" & NotaSegundoTrimestre != "PA",
         NotaTercerTrimestre != "SIN NOTA" & NotaTercerTrimestre != "PA") %>%
  select(-NotaDiciembre)

table(df$NotaPrimerTrimestre)

table(df$NotaSegundoTrimestre)

table(df$NotaTercerTrimestre)

```

```r
# prompt: ahora quiero transformar las columnas de notas en intenger

df <- df %>%
  mutate(NotaPrimerTrimestre = as.integer(NotaPrimerTrimestre),
         NotaSegundoTrimestre = as.integer(NotaSegundoTrimestre),
         NotaTercerTrimestre = as.integer(NotaTercerTrimestre))
summary(df)
```

```r
# prompt: agregar la columna promedio al df

df <- df %>%
  mutate(promedio = (NotaPrimerTrimestre + NotaSegundoTrimestre + NotaTercerTrimestre) / 3)
```

```r



table(df$Asignatura)
```

Normalizando las variables de asignatura

```r
# prompt: Ahora, Vamos a normalizar las variables de la columna asignatura, vamos a dejar solamente las materias: Matematica, Lengua, Ciencias Sociales, Ciencias Naturales, ingles, Artes visuales, Educacion fisica, Educacion Tecnologica, Educacion Musical, todos estos nombres deben ser normalizados ya que por ejemplo en ingles, esta escrito de distintos formatos, y ademas en ingles se debe englobar "taller de ingles", que tambien esta escrito en distintos formatos y borrar las que no coincidan

# Normalizar nombres de asignaturas
df <- df %>%
  mutate(Asignatura = case_when(
    grepl("matematica", Asignatura, ignore.case = TRUE) ~ "Matematica",
    grepl("lengua", Asignatura, ignore.case = TRUE) ~ "Lengua",
    grepl("ciencias sociales", Asignatura, ignore.case = TRUE) ~ "Ciencias Sociales",
    grepl("ciencias naturales", Asignatura, ignore.case = TRUE) ~ "Ciencias Naturales",
    grepl("artes visuales", Asignatura, ignore.case = TRUE) ~ "Artes Visuales",
    grepl("educacion fisica", Asignatura, ignore.case = TRUE) ~ "Educacion Fisica",
    grepl("educacion tecnologica", Asignatura, ignore.case = TRUE) ~ "Educacion Tecnologica",
    grepl("educacion musical", Asignatura, ignore.case = TRUE) ~ "Educacion Musical",
    TRUE ~ "Otro" # Categoría para asignaturas no coincidentes
  ))
# ingles no fue incluido porque primeramente se habia probado incluirlo y solo contaba con cerca de 1500 registros


# Eliminar filas donde Asignatura es "Otro"
df <- df %>% filter(Asignatura != "Otro")

table(df$Asignatura)
```

```r
library(dplyr)


df <- df %>%
  mutate(Aprobado = ifelse(NotaPrimerTrimestre == 1 | NotaSegundoTrimestre == 1 | NotaTercerTrimestre == 1, "No Aprobado", "Aprobado"))
no_aprobados <- df %>%
  filter(Aprobado == "No Aprobado") %>%
  nrow()

print(no_aprobados)




```

```r
library(dplyr)


nrow(df)
duplicados <- df %>%
  group_by(across(everything())) %>%
    filter(n() > 1) %>%
    ungroup()
df_sin_duplicados <- anti_join(df, duplicados, by = "Documento")



nrow(df_sin_duplicados)
```

Limpio los alumnos con menos o mas de 8 materias

```r
library(dplyr)


nrow(df_sin_duplicados)
# Filtrar los documentos que aparecen menos o mas de 8
documentos_a_eliminar <- df_sin_duplicados %>%
  group_by(Documento) %>%
  filter(n() != 8) %>%
  pull(Documento) %>%
  unique()

# Eliminar los registros con esos documentos
df_1_filtrado <- df_sin_duplicados %>%
  filter(!(Documento %in% documentos_a_eliminar))

nrow(df_1_filtrado)

table(df_1_filtrado$Asignatura)
```

```r
library(dplyr)

# Filtrar documentos con menos o mas de 8 registros

documentos_mas_menos_de_8 <- df_1_filtrado %>%
  group_by(Documento) %>%
  filter(n() != 8 ) %>%
  summarise(Cantidad_registros = n()) %>%  # Contar registros por documento
  filter(Cantidad_registros < 6)  # Filtrar los que tienen menos de 6





# Mostrar la cantidad de documentos filtrados
nrow(documentos_mas_menos_de_8)
```

Eliminar alumnos con mas de 1 registro por materia

```r
library(dplyr)


# Identificar documentos con más de una vez la misma materia
documentos_a_eliminar <- df_1_filtrado %>%
  group_by(Documento, Asignatura) %>%  # Agrupar por documento y asignatura
  filter(n() > 1) %>%  # Filtrar grupos con más de un registro
  pull(Documento) %>%  # Extraer los documentos
  unique()  # Obtener una lista única de documentos

# Eliminar los registros con esos documentos
df_filtrado <- df_1_filtrado %>%
  filter(!(Documento %in% documentos_a_eliminar))
```

```r
#total de alumnos por documento unico

library(dplyr)

total_alumnos <- df_1_filtrado %>%
  distinct(Documento) %>%
  nrow()

print(total_alumnos)
```
