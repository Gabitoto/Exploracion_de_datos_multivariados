```r
library("readxl")
 # Leer el archivo Excel
 nombre_excel <- "/content/df_secundario_comun.xlsx" # Changed to direct path
 df_secundario_comun_inicio <- read_excel(nombre_excel, sheet = 1)
 df_secundario_comun_inicio
```

```r
table(df_secundario_comun_inicio$curso)
```

```r
cantidad_escuelas_comunes <- length(unique(df_secundario_comun_inicio$suborganizacion))

print(cantidad_escuelas_comunes)
```

```r
library(stringr)

asignaturas_a_filtrar <- c("MATEMATICA", "LENGUA Y LITERATURA", "GEOGRAFIA", "HISTORIA",
                          "BIOLOGIA", "FISICA", "QUIMICA", "LENGUAS EXTRANJERAS",
                          "EDUCACION FISICA","LITERATURA LATINOAMERICANA","LITERATURA ARGENTINA")

# Crear una expresión regular para buscar las asignaturas
patron_asignaturas <- paste(asignaturas_a_filtrar, collapse = "|")

# Filtrar el dataframe
df_secundario_comun_inicio2 <- df_secundario_comun_inicio %>%
  filter(str_detect(Asignatura, regex(patron_asignaturas, ignore_case = TRUE)))

table(df_secundario_comun_inicio2$Asignatura)
```

```r
table(df_secundario_comun_inicio2$curso)
```

```r
df_secundario_comun_inicio2 <- df_secundario_comun_inicio2 %>%
  filter(Asignatura != "GEOGRAFIA AMBIENTAL") %>%
  mutate(Asignatura = ifelse(Asignatura %in% c("LITERATURA LATINOAMERICANA", "LITERATURA ARGENTINA"),
                            "LENGUA Y LITERATURA", Asignatura))

table(df_secundario_comun_inicio2$Asignatura)
```

```r
table(df_secundario_comun_inicio2$curso)
```

```r
library(ggplot2)

# Crear el gráfico de barras
ggplot(df_secundario_comun_inicio2, aes(x = Asignatura, fill = curso)) +
  geom_bar(position = "dodge") +  # Usar position = "dodge" para barras lado a lado
  labs(title = "Cantidad de alumnos por asignatura y división",
       x = "Asignatura",
       y = "Cantidad de alumnos",
       fill = "División") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))  # Rotar etiquetas del eje x
```

```r
df_secundario_comun_inicio3 <- df_secundario_comun_inicio2 %>%
  mutate(
    NotaPrimerTrimestre = as.numeric(NotaPrimerTrimestre),
    NotaSegundoTrimestre = as.numeric(NotaSegundoTrimestre),
    NotaTercerTrimestre = as.numeric(NotaTercerTrimestre),
    NotaDiciembre = as.numeric(NotaDiciembre)
  )
```

```r
colSums(is.na(df_secundario_comun_inicio3[,c("NotaPrimerTrimestre", "NotaSegundoTrimestre", "NotaTercerTrimestre")]))
```

```r
dim(df_secundario_comun_inicio3)
```

```r
df_secundario_comun_inicio4 <- df_secundario_comun_inicio3 %>%
  drop_na(NotaPrimerTrimestre, NotaSegundoTrimestre, NotaTercerTrimestre)
```

```r
dim(df_secundario_comun_inicio4)
```

```r
df_secundario_comun_inicio5 <- df_secundario_comun_inicio4 %>%
  mutate(Promedio = rowMeans(across(c(NotaPrimerTrimestre, NotaSegundoTrimestre, NotaTercerTrimestre)), na.rm = TRUE))
```

```r
df_secundario_comun_inicio6 <- df_secundario_comun_inicio5 %>%
  mutate(Resultado = case_when(
    Promedio >= 6 ~ "Aprobo",  # Si el promedio es mayor o igual a 6, aprueba
    NotaTercerTrimestre <= 5 & !is.na(NotaDiciembre) & NotaDiciembre >= 5 ~ "Aprobo", # Si el tercer trimestre es menor o igual a 5, tiene nota en diciembre y es mayor o igual a 5, aprueba
    NotaTercerTrimestre <= 5 & !is.na(NotaDiciembre) & NotaDiciembre < 5 ~ "No aprobo", # Si el tercer trimestre es menor o igual a 5, tiene nota en diciembre y es menor a 5, no aprueba
    TRUE ~ "No aprobo"  # En cualquier otro caso, no aprueba
  ))
```

```r
View(df_secundario_comun_inicio6)
```

### procedemos a limpiar los duplicados y normalizar los registros de materias para su analisis

```r
library(dplyr)

nrow(df_secundario_comun_inicio6)
duplicados <- df_secundario_comun_inicio6 %>%
  group_by(across(everything())) %>%
    filter(n() > 1) %>%
    ungroup()
df_sin_duplicados <- anti_join(df_secundario_comun_inicio6, duplicados, by = "Documento")



nrow(df_sin_duplicados)
```

```r
nrow(df_sin_duplicados)
# Filtrar los documentos que aparecen menos o mas de 8
documentos_a_eliminar <- df_sin_duplicados %>%
  group_by(Documento) %>%
  filter(n() == 10) %>%
  pull(Documento) %>%
  unique()

# Eliminar los registros con esos documentos
df_1_filtrado <- df_sin_duplicados %>%
  filter(!(Documento %in% documentos_a_eliminar))

nrow(df_1_filtrado)

table(df_1_filtrado$Asignatura)
```

```r
alumnos_por_curso <- df_1_filtrado %>%
  group_by(curso) %>%
  summarise(Cantidad_alumnos = n_distinct(Documento))

print(alumnos_por_curso)
```

```r
library(ggplot2)

# Crear el gráfico de barras
ggplot(df_1_filtrado, aes(x = Asignatura, fill = curso)) +
  geom_bar(position = "dodge") +  # Usar position = "dodge" para barras lado a lado
  labs(title = "Cantidad de alumnos por asignatura y división",
       x = "Asignatura",
       y = "Cantidad de alumnos",
       fill = "División") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))  # Rotar etiquetas del eje x
```

```r
library(ggplot2)
library(dplyr)

# Filtrar el dataframe para incluir solo Lengua y Matemática
df_filtrado_LM <- df_1_filtrado %>%
  filter(Asignatura %in% c("HISTORIA", "MATEMATICA"))

# Crear el gráfico de barras
ggplot(df_filtrado_LM, aes(x = Asignatura, fill = curso)) +
  geom_bar(position = "dodge") +
  labs(title = "Cantidad de alumnos en Historia y Matemática por curso",
       x = "Asignatura",
       y = "Cantidad de alumnos",
       fill = "Curso") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

Finalmente descargamos el archivo ya limpio de la secundaria comun

```r
library(dplyr)


# Identificar documentos con más de una vez la misma materia
documentos_a_eliminar <- df_1_filtrado %>%
  group_by(Documento, Asignatura) %>%  # Agrupar por documento y asignatura
  filter(n() > 1) %>%  # Filtrar grupos con más de un registro
  pull(Documento) %>%  # Extraer los documentos
  unique()  # Obtener una lista única de documentos

# Eliminar los registros con esos documentos
df_filtrado <- df_1_filtrado %>%
  filter(!(Documento %in% documentos_a_eliminar))

dim(df_filtrado)
```
