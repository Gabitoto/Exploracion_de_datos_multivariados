```r
# Calculo de estadisticos:

calcular_estadisticos <- function(df, columna) {
  # Asegurarse de que las bibliotecas necesarias estén instaladas y cargadas
  if (!require("modeest")) install.packages("modeest")
  library(modeest)
  if (!require("moments")) install.packages("moments")
  library(moments)
  if (!require("knitr")) install.packages("knitr")
  library(knitr)

  # Calcular estadísticos descriptivos
  media <- mean(df[[columna]])
  mediana <- median(df[[columna]])
  moda <- mfv(df[[columna]])
  desvio_estandar <- sd(df[[columna]])
  rango <- max(df[[columna]]) - min(df[[columna]])
  Q1 <- quantile(df[[columna]], 0.25)
  Q3 <- quantile(df[[columna]], 0.75)
  RIQ <- IQR(df[[columna]])
  cv <- (desvio_estandar / media) * 100
  asimetria <- skewness(df[[columna]])
  curtosis_valor <- kurtosis(df[[columna]])

  # Identificar outliers
  limite_inferior <- Q1 - 1.5 * RIQ
  limite_superior <- Q3 + 1.5 * RIQ
  outliers <- df[[columna]][df[[columna]] < limite_inferior | df[[columna]] > limite_superior]

  # Calcular percentil 70 y decil 8
  P70 <- quantile(df[[columna]], 0.70)
  D8 <- quantile(df[[columna]], 0.80)

  # Crear data frame con los estadísticos
  estadisticos <- data.frame(
    Media = media,
    Mediana = mediana,
    Moda = moda,
    Desvio_Estandar = desvio_estandar,
    Rango = rango,
    Q1 = Q1,
    Q3 = Q3,
    RIQ = RIQ,
    CV = cv,
    Asimetria = asimetria,
    Curtosis = curtosis_valor,
    P70 = P70,
    D8 = D8,
    Outliers = paste(head(round(outliers, 1), 5), collapse = ", ")
  )

  # Mostrar la tabla de estadísticos con kable()
  kable(estadisticos)
}
```

### Matematicas 1er año

```r
# 1. Definir los intervalos (bins)
breaks <- seq(1, 10, by = 1) # Intervalos de 1 en 1 desde 1 hasta 10

# 2. Crear una nueva columna con los intervalos
datos_matematicas1 <- datos_matematicas_1o %>%
  mutate(intervalo_promedio = cut(Promedio, breaks = breaks, include.lowest = TRUE,right = FALSE))

# 3. Crear la tabla de frecuencias por intervalos
tabla_frecuencias_matematicas1 <- datos_matematicas1 %>%
  group_by(intervalo_promedio) %>%
  summarise(Frecuencia_Absoluta = n()) %>%
  mutate(
    Frecuencia_Relativa = Frecuencia_Absoluta / sum(Frecuencia_Absoluta),
    Frecuencia_Porcentual = Frecuencia_Relativa * 100, # Frecuencia porcentual
    Frecuencia_Absoluta_Acumulada = cumsum(Frecuencia_Absoluta), # Acumulada absoluta
    Frecuencia_Relativa_Acumulada = cumsum(Frecuencia_Relativa), # Acumulada relativa
    Frecuencia_Porcentual_Acumulada = cumsum(Frecuencia_Porcentual) # Acumulada porcentual
  )

# 4. Mostrar la tabla de frecuencias por intervalos
View(tabla_frecuencias_matematicas1)
```
