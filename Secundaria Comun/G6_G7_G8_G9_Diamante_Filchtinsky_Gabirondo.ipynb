### ¿Podría afirmar que la media de su departamento en las notas Matemática de primer año son menores o iguales 7?


- La variable aleatoria a estudiar es "promedio" de "1er año Secundaria".

---

	H0: u <= 7
	Ha: u > 7

---
- La Hipotesis nula establece que la media de notas de matematica de 1er año son menores o iguales a 7 siete mientras que la alternativa la refuta.

- La prueba que se va a usar consiste en un Test de Wilcoxon.

- Es una prueba de cola derecha. (unilateral derecha)

- Los datos no cumplen con los supuestos de normalidad por lo cual la prueba a usar será no paramétrica.



```r
# Codigo de respuestas:

# Test de normalidad (Shapiro-Wilk)
shapiro_test1 <- shapiro.test(datos_matematicas_1o$Promedio)
print(shapiro_test1)

# Determinar si se usa Wilcoxon o prueba t
if (shapiro_test1$p.value >= 0.05) {
  cat("\nLos datos siguen una distribución normal (p ≥ 0.05). Se usaría prueba t.\n")
  usar_t_test <- TRUE
} else {
  cat("\nLos datos NO siguen una distribución normal (p < 0.05). Se usará Wilcoxon.\n")
  usar_t_test <- FALSE
}

# Prueba de Wilcoxon para la mediana poblacional
prueba_mediana1 <- wilcox.test(datos_matematicas_1o$Promedio, mu = 7, alternative = "greater")
print(prueba_mediana1)

# Conclusión
if (prueba_mediana1$p.value < 0.05) {
  cat("\nConclusión: Rechazamos H0, la mediana poblacional podria ser mayor a 7.\n")
} else {
  cat("\nConclusión: No podemos rechazar H0, la mediana poblacional podria ser menor o igual 7.\n")
}

# Graficar Boxplot para visualizar distribución y posibles valores atípicos
boxplot(datos_matematicas_1o$Promedio,
        main = "Distribución de Promedios de Matematicas en 1er año",
        ylab = "Notas",
        col = "lightblue",
        border = "black")

# Agregar línea de la mediana esperada
abline(h = 7, col = "red", lty = 2)

# Agregar la media al boxplot
points(mean(datos_matematicas_1o$Promedio), col = "blue", pch = 19, cex = 1.5)

```

- El p valor obtenido es de: p-value = 0.124

- Por lo cual se concluye que: No podemos rechazar H0, la mediana poblacional podria ser menor o igual 7.

### Es cierto que, en su departamento, en todos los años del sistema educativo, ¿las notas que se obtienen en el tercer trimestre de Historia son menores o iguales a las del primer trimestre de la misma materia?

- La variable aleatoria a estudiar es "NotasPrimerTrimestre" y "NotasTercerTrimestre" de mi departamento en la materia Historia.

---

	H0: U3 <= U1
	Ha: U3 > U1

---
- La Hipotesis nula establece que la media de notas de TercerTrimestre de mi departamento en todos los años son menores o iguales a las notas de primer trimestre, mientras que la hipotesis alternativa difiere de esto estableciendo que son mayores.

- La prueba que se va a usar consiste en un Test de Wilcoxon ya que son datos pareados.

- Es una prueba de cola derecha. (unilateral derecha)

- Los datos no cumplen con los supuestos de normalidad por lo cual la prueba a usar será no paramétrica.

```r
# Instalar paquetes si es necesario
if (!require(car)) install.packages("car", dependencies=TRUE)

# Cargar la librería necesaria
library(car)

# Historia 1er año:

# Calcular las diferencias entre las notas del tercer y primer trimestre para cada estudiante
diferencias1 <- datos_historia_1o$NotaTercerTrimestre - datos_historia_1o$NotaPrimerTrimestre

# 1. Test de Normalidad: Kolmogorov-Smirnov
ks_test1a <- ks.test(diferencias1, "pnorm")
print(ks_test1a)

# Decidir qué prueba usar según la normalidad
if (ks_test1a$p.value >= 0.05) {
  cat("\nLos datos siguen una distribución normal (p ≥ 0.05). Se usa la prueba t pareada.\n")
  usar_t_test <- TRUE
} else {
  cat("\nLos datos NO siguen una distribución normal (p < 0.05). Se usará la prueba de Wilcoxon.\n")
  usar_t_test <- FALSE
}

# 2. Prueba si no son normales:
prueba1 <- wilcox.test(datos_historia_1o$NotaTercerTrimestre, datos_historia_1o$NotaPrimerTrimestre, paired = TRUE, alternative = "greater")
prueba1
# Interpretación del resultado
if (prueba1$p.value < 0.05) {
  cat("\nConclusión: Rechazamos H0. Por lo que las notas de tercer trimestre son mayores a las del primer trimestre.\n")
} else {
  cat("\nConclusión: No podemos rechazar H0. No hay evidencia suficiente para rechazar que las notas del tercer trimestres son menores que las del primer trimestre.\n")
}

# Graficar Boxplot comparativo
boxplot(datos_historia_1o$NotaTercerTrimestre, datos_historia_1o$NotaPrimerTrimestre,
        names = c("Tercer trimestre", "Primer Trimestre"),
        main = "Comparación de notas",
        ylab = "Notas sacadas",
        col = c("lightblue", "lightgreen"),
        border = "black")

# Agregar líneas de medias
points(1, mean(datos_historia_1o$NotaTercerTrimestre), col = "red", pch = 19, cex = 1.5)
points(2, mean(datos_historia_1o$NotaPrimerTrimestre), col = "blue", pch = 19, cex = 1.5)
```

### ¿Podría afirmar que la media en las notas de Geografia de primer año estuvo por encima de 7?

- La variable aleatoria a estudiar es "promedio" de "1er año Secundaria" de la materia Geografia.

---

	H0: u <= 7
	Ha: u > 7

---
- La Hipotesis nula establece que la media de notas de Geografia de 1er año son menores o iguales a 7 siete mientras que la alternativa la refuta, para poder contestar la pregunta de hipotesis debemos rechazar la hipotesis nula.

- La prueba que se va a usar consiste en un Test de Wilcoxon.

- Es una prueba de cola derecha. (unilateral derecha)

- Los datos no cumplen con los supuestos de normalidad por lo cual la prueba a usar será no paramétrica.

```r
# Codigo de respuestas:

# 1. Test de Normalidad: Kolmogorov-Smirnov
ks_test1 <- ks.test(datos_geografia_1o$Promedio, "pnorm")
print(ks_test1)

# Determinar si se usa Wilcoxon o prueba t
if (ks_test1$p.value >= 0.05) {
  cat("\nLos datos siguen una distribución normal (p ≥ 0.05). Se usaría prueba t.\n")
  usar_t_test <- TRUE
} else {
  cat("\nLos datos NO siguen una distribución normal (p < 0.05). Se usará Wilcoxon.\n")
  usar_t_test <- FALSE
}

# Prueba de Wilcoxon para la mediana poblacional
prueba_mediana1g <- wilcox.test(datos_geografia_1o$Promedio, mu = 7, alternative = "greater")
print(prueba_mediana1g)

# Conclusión
if (prueba_mediana1g$p.value < 0.05) {
  cat("\nConclusión: Rechazamos H0, la mediana poblacional podria ser mayor a 7.\n")
} else {
  cat("\nConclusión: No podemos rechazar H0, la mediana poblacional podria ser menor o igual 7.\n")
}

# Graficar Boxplot para visualizar distribución y posibles valores atípicos
boxplot(datos_geografia_1o$Promedio,
        main = "Distribución de Promedios de Geografia en 1er año",
        ylab = "Notas",
        col = "lightblue",
        border = "black")

# Agregar línea de la mediana esperada
abline(h = 7, col = "red", lty = 2)

# Agregar la media al boxplot
points(mean(datos_geografia_1o$Promedio), col = "blue", pch = 19, cex = 1.5)
```

### En su departamento, ¿hay diferencias significativas entre el rendimiento en Matemática de primer año de escuelas secundarias Publicas y Privadas?



- La variable aleatoria a estudiar es "promedio" de "1er año Secundaria" de la materia Matematicas de escuelas publicas y privadas.

---

	H0: Upub = Upri
	Ha: Upub != Upri

---
- La Hipotesis nula establece que el rendimiento de promedios de Matematicas de 1er año de Publicas y Privadas son iguales mientras que la alternativa la refuta, evidenciando que son diferentes.

- La prueba que se va a usar consiste en un Test de Wilcoxon.

- Es una prueba de dos colas. (bilateral)

- Los datos no cumplen con los supuestos de normalidad por lo cual la prueba a usar será no paramétrica.

```r
# Filtrar para obtener datos de escuelas privadas
datos_matematicas_1o_privada <- datos_matematicas_1o %>%
  filter(EsPrivada == "S")

# Filtrar para obtener datos de escuelas públicas
datos_matematicas_1o_publica <- datos_matematicas_1o %>%
  filter(EsPrivada == "N")
```

```r
# Datos
matematicas_1ro_privada <- datos_matematicas_1o_privada$Promedio
matematicas_1ro_publica <- datos_matematicas_1o_publica$Promedio

# Crear un data frame con los datos
data_pp <- data.frame(
  Datos = c(matematicas_1ro_privada, matematicas_1ro_publica),
  grupo = rep(c("matematicas_1ro_privada", "matematicas_1ro_publica"), times = c(length(matematicas_1ro_privada), length(matematicas_1ro_publica)))
)

# 1. Test de Normalidad: KS
# Considera inspeccionar visualmente la distribución con un histograma o gráfico Q-Q.

ks_test_privada <- ks.test(matematicas_1ro_privada, "pnorm")
print("K-S para datos de escuelas privadas:")
print(ks_test_privada)

ks_test_publica <- ks.test(matematicas_1ro_publica, "pnorm")
print("K-S para datos de escuelas publicas:")
print(ks_test_publica)


# 2. Test de Levene para igualdad de varianzas
# Asegúrate de tener cargada la librería 'car'
# install.packages("car", dependencies=TRUE)
# library(car)
test_levene_pp <- leveneTest(Datos ~ grupo, data = data_pp)

# Imprimir resultado del test
print("\nTest de Levene para igualdad de varianzas:")
print(test_levene_pp)

# Extraer p-valor
p_valor_levene <- test_levene_pp$`Pr(>F)`[1]

# Determinar la conclusión para el test de Levene
if (p_valor_levene < 0.05) {
  cat("\nConclusión Levene:", "Las varianzas son significativamente diferentes (p < 0.05). Se recomienda utilizar una prueba no paramétrica o una prueba t de Welch (si los datos fueran normales).\n")
  varianzas_iguales <- FALSE
} else {
  cat("\nConclusión Levene:", "No hay evidencia suficiente para afirmar que las varianzas son diferentes (p ≥ 0.05). Se puede asumir homogeneidad de varianza.\n")
  varianzas_iguales <- TRUE
}

# Crear un boxplot comparativo
boxplot(Datos ~ grupo, data = data_pp,
        col = c("lightblue", "lightgreen"),
        main = "Promedios en Matemática de 1er año (Públicas vs Privadas)",
        xlab = "Tipo de Escuela",
        ylab = "Promedio",
        border = "black")

# Agregar líneas de medias
points(1, mean(matematicas_1ro_privada), col = "red", pch = 19, cex = 1.5)
points(2, mean(matematicas_1ro_publica), col = "blue", pch = 19, cex = 1.5)


# 3. Prueba de hipótesis adecuada (Wilcoxon-Mann-Whitney)
# la prueba de Wilcoxon-Mann-Whitney es la más adecuada.
wilcox_test_publica_privada <- wilcox.test(matematicas_1ro_privada, matematicas_1ro_publica, alternative = "two.sided")
print("\nResultado de la prueba de Wilcoxon-Mann-Whitney:")
print(wilcox_test_publica_privada)

# Conclusión de la prueba de Wilcoxon
if (wilcox_test_publica_privada$p.value < 0.05) {
  cat("\nConclusión Wilcoxon: Rechazamos H0. Hay una diferencia significativa en el rendimiento en Matemática de primer año entre escuelas públicas y privadas.\n")
} else {
  cat("\nConclusión Wilcoxon: No podemos rechazar H0. No hay evidencia suficiente para afirmar que hay una diferencia significativa en el rendimiento en Matemática de primer año entre escuelas públicas y privadas.\n")
}
```

- El nivel de confianza es de: 95%

- El valor del estadístico de la prueba es: W = 47318

- El p valor obtenido es de: p-value = 0.0003305

- Por lo cual se concluye que: Rechazamos H0. Hay una diferencia significativa en el rendimiento en Matemática de primer año entre escuelas públicas y privadas.

```r
require(dplyr)
require(tidyr)

# Agrupar por escuela y curso, y contar el número de alumnos únicos por curso en cada escuela
matricula_por_escuela_curso <- datos_secundaria_comun %>%
  group_by(CUE, curso) %>% # Agrupar por código de establecimiento y curso
  summarise(Matricula = n_distinct(Documento), .groups = 'drop') # Contar alumnos únicos

# Definir los años para cada ciclo
años_basico <- c("1ro Secundaria", "2do Secundaria", "3ro Secundaria")
años_orientado <- c("4to Secundaria", "5to Secundaria", "6to Secundaria")

# Pivotear para tener los cursos como columnas
matricula_escuela_wide <- matricula_por_escuela_curso %>%
  pivot_wider(names_from = curso, values_from = Matricula, values_fill = 0) # Usar values_fill = 0 para escuelas sin alumnos en un curso específico

# Sumar la matrícula por ciclo para cada escuela
matricula_por_ciclo_escuela <- matricula_escuela_wide %>%
  mutate(
    Matricula_Basico = select(., any_of(años_basico)) %>% rowSums(na.rm = TRUE), # Sumar matrículas de años del ciclo básico
    Matricula_Orientado = select(., any_of(años_orientado)) %>% rowSums(na.rm = TRUE) # Sumar matrículas de años del ciclo orientado
  ) %>%
  select(CUE, Matricula_Basico, Matricula_Orientado) # Seleccionar solo las columnas relevantes

# Mostrar las primeras filas del nuevo dataframe
print(head(matricula_por_ciclo_escuela))

## Verifica si hay datos en ambos ciclos a nivel de escuela
if (nrow(matricula_por_ciclo_escuela) == 0) {
  stop("No hay suficientes datos de matrícula por escuela para ambos ciclos en este departamento.")
}

# Extraer las matrículas de los ciclos como vectores para las pruebas estadísticas
matricula_basico_escuela <- matricula_por_ciclo_escuela$Matricula_Basico
matricula_orientado_escuela <- matricula_por_ciclo_escuela$Matricula_Orientado


# Verifica si hay datos en ambos ciclos
if (length(matricula_basico_escuela) == 0 || length(matricula_orientado_escuela) == 0) {
  stop("No hay suficientes datos de matrícula para ambos ciclos en este departamento.")
}

# Ejemplo con Kolmogorov-Smirnov
ks_test_basico_escuela <- ks.test(matricula_basico_escuela, "pnorm", mean(matricula_basico_escuela), sd(matricula_basico_escuela))
print("K-S para la Matrícula del Ciclo Básico por Escuela:")
print(ks_test_basico_escuela)

# Test de Normalidad para la matrícula del ciclo orientado por escuela
ks_test_orientado_escuela <- ks.test(matricula_orientado_escuela, "pnorm", mean(matricula_orientado_escuela), sd(matricula_orientado_escuela))
print("K-S para la Matrícula del Ciclo Orientado por Escuela:")
print(ks_test_orientado_escuela)

# Determinar si usar prueba t o Wilcoxon
usar_t_test <- FALSE
if (ks_test_basico_escuela$p.value >= 0.05 && ks_test_orientado_escuela$p.value >= 0.05) {
  cat("\nLos datos de ambos ciclos siguen una distribución normal (p ≥ 0.05). Se usará prueba t.\n")
  usar_t_test <- TRUE
} else {
  cat("\nAl menos uno de los grupos NO sigue una distribución normal (p < 0.05). Se usará Wilcoxon.\n")
}


# 4. Realizar la Prueba Estadística

if (usar_t_test) {
  # Test de Levene para igualdad de varianzas
  # Asegúrate de tener cargada la librería 'car'
  if (!require(car)) install.packages("car", dependencies=TRUE)
  library(car)

  data_matricula <- data.frame(
    Matricula = c(matricula_basico_escuela, matricula_orientado_escuela),
    Ciclo = rep(c("Basico", "Orientado"), times = c(length(matricula_basico_escuela), length(matricula_orientado_escuela)))
  )

  test_levene <- leveneTest(Matricula ~ Ciclo, data = data_matricula)
  print("\nResultado del Test de Levene para igualdad de varianzas:")
  print(test_levene)

  # Realizar la prueba t
  if (test_levene$`Pr(>F)`[1] >= 0.05) {
    cat("\nLas varianzas son homogéneas. Realizando prueba t asumiendo igualdad de varianzas.\n")
    t_test_resultado <- t.test(matricula_basico_escuela, matricula_orientado_escuela, var.equal = TRUE, alternative = "two.sided")
  } else {
    cat("\nLas varianzas son heterogéneas. Realizando prueba t de Welch.\n")
    t_test_resultado <- t.test(matricula_basico_escuela, matricula_orientado_escuela, var.equal = FALSE, alternative = "two.sided")
  }
  print("\nResultado de la prueba t:")
  print(t_test_resultado)
  p_valor_final <- t_test_resultado$p.value

} else {
  # Realizar la prueba de Wilcoxon-Mann-Whitney
  wilcox_test_resultado <- wilcox.test(matricula_basico_escuela, matricula_orientado_escuela, alternative = "two.sided")
  print("\nResultado de la prueba de Wilcoxon-Mann-Whitney:")
  print(wilcox_test_resultado)
  p_valor_final <- wilcox_test_resultado$p.value
}

cat("\n--- Conclusión del Test Estadístico ---\n")
if (p_valor_final < 0.05) {
  cat("Con un nivel de significancia del 5%, rechazamos la hipótesis nula.\n")
  cat("Hay evidencia estadística suficiente para afirmar que la matrícula del ciclo básico difiere significativamente de la del ciclo orientado en el departamento Diamante.\n")
} else {
  cat("Con un nivel de significancia del 5%, no podemos rechazar la hipótesis nula.\n")
  cat("No hay evidencia estadística suficiente para afirmar que la matrícula del ciclo básico difiere significativamente de la del ciclo orientado en el departamento Diamante.\n")
}

# Crear un data frame para el boxplot
data_matricula_plot <- data.frame(
    Matricula = c(matricula_basico_escuela, matricula_orientado_escuela),
    Ciclo = rep(c("Basico", "Orientado"), times = c(length(matricula_basico_escuela), length(matricula_orientado_escuela)))
  )

boxplot(Matricula ~ Ciclo, data = data_matricula_plot,
        main = paste("Comparación de Matrícula por Ciclo en"),
        ylab = "Número de Matrícula (por curso/año)",
        col = c("lightblue", "lightgreen"),
        border = "black")

# Opcional: Agregar líneas de medias o medianas
points(1, mean(matricula_basico_escuela), col = "red", pch = 19, cex = 1.5) # Media del básico
points(2, mean(matricula_orientado_escuela), col = "blue", pch = 19, cex = 1.5) # Media del orientado

# O para medianas si usaste Wilcoxon:
# points(1, median(matricula_basico), col = "red", pch = 19, cex = 1.5)
# points(2, median(matricula_orientado), col = "blue", pch = 19, cex = 1.5)
```

- Para probar que el test es confiables mediremos su potencia

```r
install.packages("pwr", dependencies = TRUE)
install.packages("effsize", dependencies = TRUE)
```

```r
# Cargar las librerías
library(pwr)
library(effsize)
install.packages("lsr")
# Cargar la librería 'lsr'
library(lsr)
```

```r
# Asumiendo que matricula_basico_escuela y matricula_orientado_escuela son tus vectores de datos
# obtenidos previamente, donde cada elemento representa la matrícula total para una escuela
# en el ciclo respectivo.

# Calcular el tamaño del efecto (d de Cohen)
# Asegúrate de que los dos vectores tengan la misma longitud antes de calcular la d de Cohen
if (length(matricula_basico_escuela) == length(matricula_orientado_escuela)) {
  cohen_d <- cohensD(matricula_basico_escuela, matricula_orientado_escuela)
  cat("d de Cohen calculada:", cohen_d, "\n")

  # Tamaño de la muestra por grupo (número de escuelas)
  n_per_group_ttest <- length(matricula_basico_escuela)
  cat("Tamaño de la muestra por grupo:", n_per_group_ttest, "\n")

  # Realizar el análisis de potencia para una prueba t de dos muestras
  power_analysis_ttest <- pwr.t.test(
    n = n_per_group_ttest,       # Tamaño de la muestra por grupo
    d = cohen_d,                # Tamaño del efecto (d de Cohen)
    sig.level = 0.05,           # Nivel de significancia
    type = "two.sample",        # Prueba t de dos muestras
    alternative = "two.sided"   # Prueba bilateral (basada en tu hipótesis)
  )

  # Imprimir los resultados del análisis de potencia
  print("\nAnálisis de potencia para la prueba t de dos muestras:")
  print(power_analysis_ttest)

} else {
  cat("Error: La longitud de 'matricula_basico_escuela' y 'matricula_orientado_escuela' no son iguales.\n")
}
# Interpretar la potencia
if (power_analysis_ttest$power < 0.80) {
  cat("\nConclusión sobre la Potencia (T test): La potencia calculada (", round(power_analysis_ttest$power, 3), ") es baja (menos de 0.80).\n",
      "Dado el tamaño de muestra muy pequeño (n1=", power_analysis_ttest$n, ", n2=", power_analysis_ttest$n, ") y el tamaño del efecto estimado (d=", round(power_analysis_ttest$d, 3), "),\n",
      "este test tiene una baja probabilidad de detectar una diferencia significativa si existiera una del tamaño estimado.\n",
      "Un resultado no significativo es probable debido a la falta de potencia.")
} else {
  cat("\nConclusión sobre la Potencia (Wilcoxon): La potencia calculada (", round(power_analysis_ttest$power, 3), ") es suficiente (0.80 o mayor).\n",
      "Tienes una buena probabilidad de detectar el tamaño del efecto estimado si este fuera real con tu tamaño de muestra actual.")
}
```

```r
# --- Parámetros para el cálculo del tamaño de muestra ---

# Potencia deseada (ej., 0.80 para un 80% de potencia)
potencia_deseada <- 0.80

# Nivel de significancia (alfa)
nivel_significancia <- 0.05

# Tamaño del efecto estimado (d de Cohen)
# Debes reemplazar este valor con tu tamaño del efecto estimado.
# Por ejemplo, si quieres detectar un tamaño del efecto medio (d=0.5):
tamaño_efecto_estimado <- 0.22
# O si utilizas la d de Cohen calculada a partir de tus datos actuales (úsalo con precaución):
# tamaño_efecto_estimado <- cohen_d # asumiendo que cohen_d se calculó en el paso anterior

# --- Calcular el tamaño de muestra requerido ---

resultado_tamaño_muestra <- pwr.t.test(
  d = tamaño_efecto_estimado,  # Tamaño del efecto estimado
  power = potencia_deseada,      # Potencia deseada
  sig.level = nivel_significancia,  # Nivel de significancia
  type = "two.sample",      # Prueba t de dos muestras
  alternative = "two.sided" # Prueba bilateral (basada en tu hipótesis)
)

# Imprimir el resultado
print("Tamaño de muestra requerido por grupo para la prueba t:")
print(resultado_tamaño_muestra)
```

#### Realiza un test F (ANOVA) para comparar las medias de las poblaciones. ¿Cuáles serían las hipótesis nula y alternativa?

- La variable aleatoria a estudiar es "Promedio" de todos los años de la materia Matematica y luego Lengua.

---

	H0: μ_1 = μ_2 = μ_3 = μ_4 = μ_5 = μ_6
	Ha: μ_1 ≠ μ_2 ≠ μ_3 ≠ μ_4 ≠ μ_5 ≠ μ_6

---
- La Hipotesis nula establece que la media de las materias entre años son iguales mientras que la alternativa la refuta.

- La prueba que se va a usar consiste en un Test de ANOVA de un factor.

- Los datos no cumplen con los supuestos de normalidad por lo cual la prueba a usar será no paramétrica.

```r
# Realizar el test ANOVA
# La fórmula indica que queremos modelar 'Promedio' en función de 'curso'
anova_resultado <- aov(Promedio ~ curso, data = datos_matematicas_todos_años)

# Mostrar el resumen del ANOVA
summary(anova_resultado)
```

- Tras evaluar la tabla ANOVA, ¿cuál sería tu conclusión en el contexto del problema?

- Dado que el p-valor (3.07e-12) es mucho menor que el nivel de significancia común de 0.05 (o incluso 0.01 o 0.001), rechazamos la hipótesis nula (H0).

- Esto significa que hay evidencia estadísticamente significativa, con un alto grado de confianza, para concluir que al menos una de las medias de los promedios de Matemática entre los diferentes años de secundaria es significativamente diferente de las otras.

#### Si se han obtenido diferencias significativas entre los grupos, determina dónde se muestran esas diferencias utilizando el test HSD de Tukey.

```r
# Instalar el paquete 'lsr' si no está instalado
install.packages("lsr")
# Cargar la librería 'lsr'
library(lsr)

#Calcular f para anova

eta <- etaSquared(anova_resultado, anova = TRUE)["curso", "eta.sq"]

f <- sqrt(eta / (1 - eta))
print(f)
pwr.anova.test(k = 6 , n = 408 , f , sig.level =0.05 )   # k cantidad de grupos, n el menor número de datos entre los grupos

# Realizar la prueba HSD de Tukey
tukey_resultado <- TukeyHSD(anova_resultado, "curso", conf.level = .95)

# Mostrar el resumen de los resultados de Tukey HSD
print(tukey_resultado)
```

```r
# Potencia
if (!require(pwr)) install.packages("pwr")
# Cargar la librería
library(pwr)

eta2 <- resultado$ANOVA$ges[1]          # ges = eta² generalizado

# Calcular eta cuadrado parcial
f_value <- sqrt(eta2 / (1 - eta2))


# Calcular potencia
# k = 3 condiciones (tiempos), n = 579 sujetos, f  (efecto grande), sig.level = 0.05
pwr.anova.test(k = 3, n = 579, f = f_value, sig.level = 0.05)
```

- No encuentra diferencias significativas entre las variables

Por lo que procederemos a hacer test estadisticos para corroborar estas diferencias, plantenado nuestra hipotesis como:

--- Formulación de Hipótesis ---

Para el Efecto Principal del Año:

-  H0: Las medias de las notas de matemática son iguales en todos los años de secundaria (μ1 = μ2 = ... = μk).

-  H1: Al menos una de las medias de las notas de matemática difiere entre los años.

-  H1: Las medias de las notas de matemática difieren entre el sector público y privado.

Para el Efecto de Interacción (Año x Sector):

-  H0: No hay interacción entre el año de secundaria y el sector en cuanto a las notas de matemática (el efecto del año no depende del sector, y viceversa).

-  H1: Existe interacción entre el año de secundaria y el sector en cuanto a las notas de matemática.


```r
# Asegúrate de tener la librería dplyr cargada
library(dplyr)

# Crear el nuevo dataset con las columnas seleccionadas
datos_anova <- datos_matematicas_1 %>%
  select(Promedio, curso, EsPrivada)

# Opcional: Ver la estructura del nuevo dataset
str(datos_anova)

# Opcional: Ver las primeras filas del nuevo dataset
head(datos_anova)
```

```r
# Asegurarse de que Anio y Sector sean factores (si no lo son por defecto al cargar)
datos_anova$curso <- as.factor(datos_anova$curso)
datos_anova$EsPrivada <- as.factor(datos_anova$EsPrivada)

# --- Paso 1: Formulación de Hipótesis ---
cat("\n--- Formulación de Hipótesis ---\n")
cat("Para el Efecto Principal del Año:\n")
cat("  H0: Las medias de las notas de matemática son iguales en todos los años de secundaria (μ1 = μ2 = ... = μk).\n")
cat("  H1: Al menos una de las medias de las notas de matemática difiere entre los años.\n")

cat("\nPara el Efecto Principal del Sector:\n")
cat("  H0: Las medias de las notas de matemática son iguales en el sector público y privado (μpublico = μprivado).\n")
cat("  H1: Las medias de las notas de matemática difieren entre el sector público y privado.\n")

cat("\nPara el Efecto de Interacción (Año x Sector):\n")
cat("  H0: No hay interacción entre el año de secundaria y el sector en cuanto a las notas de matemática (el efecto del año no depende del sector, y viceversa).\n")
cat("  H1: Existe interacción entre el año de secundaria y el sector en cuanto a las notas de matemática.\n")


# --- Paso 2: Verificar las Suposiciones del ANOVA ---

# Instalación y carga de librerías necesarias para supuestos y post-hoc
install.packages("car")    # Para la prueba de Levene
install.packages("ggpubr") # Para el gráfico QQ y fácil visualización de distribuciones
install.packages("emmeans") # Para análisis de efectos simples y comparaciones post-hoc

library(car)    # Para la prueba de Levene
library(ggpubr) # Para el gráfico QQ y fácil visualización de distribuciones
library(emmeans) # Para análisis de efectos simples y comparaciones post-hoc

cat("\n--- Verificación de Suposiciones del ANOVA ---\n")

# 2.1 Homogeneidad de Varianzas (Prueba de Levene)
# La función leveneTest requiere el modelo ajustado o los datos.
# Usaremos la fórmula directamente para la prueba de Levene para todos los grupos.
levene_test_result <- leveneTest(Promedio ~ curso * EsPrivada, data = datos_anova)
cat("\nPrueba de Levene para Homogeneidad de Varianzas:\n")
print(levene_test_result)
if (levene_test_result$`Pr(>F)`[1] < 0.05) {
  cat("  ¡Advertencia! El valor p < 0.05. Se rechaza H0, lo que sugiere que las varianzas NO son homogéneas. Esto podría afectar la validez del ANOVA. Considera transformaciones o un ANOVA de Welch.\n")
} else {
  cat("  El valor p >= 0.05. No hay evidencia para rechazar H0. Las varianzas son homogéneas.\n")
}

# 2.2 Normalidad de los Residuos
# Primero ajustamos el modelo ANOVA para obtener los residuos.
model_anova <- aov(Promedio ~ curso * EsPrivada, data = datos_anova)
residuals <- residuals(model_anova)

# Gráfico Q-Q de los residuos
cat("\nGráfico Q-Q para Normalidad de Residuos (verifica visualmente si los puntos siguen la línea diagonal):\n")
qqPlot(residuals, main = "Q-Q Plot de Residuos", ylab = "Cuantiles Muestrales", xlab = "Cuantiles Teóricos Normales")

# Prueba de Shapiro-Wilk sobre los residuos (para muestras pequeñas-medianas)
shapiro_test_result <- shapiro.test(residuals)
cat("\nPrueba de Shapiro-Wilk para Normalidad de Residuos:\n")
cat(paste0("  Estadístico W: ", round(shapiro_test_result$statistic, 3), "\n"))
cat(paste0("  Valor p: ", round(shapiro_test_result$p.value, 3), "\n"))
if (shapiro_test_result$p.value < 0.05) {
  cat("  ¡Advertencia! El valor p < 0.05. Se rechaza H0, lo que sugiere que los residuos NO son normales. Esto podría afectar la validez del ANOVA. Considera transformaciones o un ANOVA no paramétrico si las desviaciones son severas y la muestra es pequeña.\n")
} else {
  cat("  El valor p >= 0.05. No hay evidencia para rechazar H0. Los residuos son normales.\n")
}

# 2.3 Independencia de las Observaciones (Depende del diseño del estudio)
cat("\nIndependencia de las Observaciones: Esta suposición se evalúa en función del diseño de tu estudio. Asegúrate de que las notas de un estudiante no influyen en las de otro.\n")


# --- Paso 3: Elegir el Nivel de Significancia (Ya definido como 0.05) ---
alpha <- 0.05
cat(paste0("\n--- Nivel de Significancia (alpha): ", alpha, " ---\n"))

# --- Paso 4: Realizar el ANOVA de Dos Factores ---
# Ya ajustamos el modelo arriba: model_anova <- aov(NotaMatematica ~ Anio * Sector, data = df)
cat("\n--- Tabla de Resultados del ANOVA de Dos Factores ---\n")
print(summary(model_anova))

# --- Paso 5: Interpretar los Resultados del ANOVA ---
cat("\n--- Interpretación de los Resultados del ANOVA ---\n")

# Obtener los p-valores de la tabla summary
anova_summary <- summary(model_anova)
p_interaccion <- anova_summary[[1]]$`Pr(>F)`[3] # P-valor para Anio:Sector
p_anio <- anova_summary[[1]]$`Pr(>F)`[1]       # P-valor para Anio
p_sector <- anova_summary[[1]]$`Pr(>F)`[2]     # P-valor para Sector

cat(paste0("\nValor p para la Interacción (Anio:Sector): ", round(p_interaccion, 4), "\n"))
if (p_interaccion < alpha) {
  cat("  ¡La interacción entre Año y Sector es SIGNIFICATIVA (p < ", alpha, ")!\n")
  cat("  Esto significa que el efecto de un factor en las notas de matemática DEPENDE del nivel del otro factor.\n")
  cat("  No se deben interpretar los efectos principales de Año o Sector de forma aislada.\n")
  cat("  Proceder con el análisis de efectos simples (comparaciones dentro de cada nivel de los factores).\n")

  # Visualización de la interacción (plot de perfiles de medias)
  cat("\nVisualización de la Interacción (Gráfico de Perfiles de Medias):\n")
  # Utilizar 'ggplot2' para un gráfico de interacción más estético
  install.packages("ggplot2")
  library(ggplot2)
  interaction_plot <- ggplot(datos_anova, aes(x = curso, y = Promedio, color = EsPrivada, group = EsPrivada)) +
    stat_summary(fun = mean, geom = "line", aes(linetype = EsPrivada)) +
    stat_summary(fun = mean, geom = "point", size = 3) +
    labs(title = "Gráfico de Interacción de Medias de Notas de Matemática",
         x = "Año de Secundaria",
         y = "Media de Nota de Matemática") +
    theme_minimal()
  print(interaction_plot)
  cat("  Observe las líneas en el gráfico: si son paralelas, no hay interacción. Si se cruzan o divergen, hay interacción.\n")


  # Análisis de Efectos Simples (si la interacción es significativa)
  cat("\n--- Análisis de Efectos Simples (debido a interacción significativa) ---\n")

  cat("\nComparaciones de Sector DENTRO de cada Año:\n")
  emmeans_sector_by_anio <- emmeans(model_anova, ~ EsPrivada | curso)
  print(summary(emmeans_sector_by_anio, adjust = "tukey")) # Ajuste de Tukey para comparaciones múltiples

  cat("\nComparaciones de Año DENTRO de cada Sector:\n")
  emmeans_anio_by_sector <- emmeans(model_anova, ~ curso | EsPrivada)
  print(summary(emmeans_anio_by_anio, adjust = "tukey")) # Ajuste de Tukey para comparaciones múltiples

} else {
  cat("  La interacción entre Año y Sector NO es significativa (p >= ", alpha, ").\n")
  cat("  Esto significa que el efecto de un factor en las notas de matemática es consistente a través de los niveles del otro factor.\n")
  cat("  Se pueden interpretar los efectos principales de Año y Sector de forma independiente.\n")

  cat(paste0("\nValor p para el Efecto Principal del Año: ", round(p_anio, 4), "\n"))
  if (p_anio < alpha) {
    cat("  El Efecto Principal del Año es SIGNIFICATIVO (p < ", alpha, ").\n")
    cat("  Existen diferencias significativas en las notas de matemática entre los años de secundaria.\n")
    cat("  --- Pruebas Post-Hoc para el Año (Comparaciones por pares) ---\n")
    posthoc_anio <- TukeyHSD(model_anova, which = "curso")
    print(posthoc_anio)
  } else {
    cat("  El Efecto Principal del Año NO es significativo (p >= ", alpha, ").\n")
    cat("  No hay evidencia suficiente para concluir que existen diferencias significativas en las notas de matemática entre los años.\n")
  }

  cat(paste0("\nValor p para el Efecto Principal del Sector: ", round(p_sector, 4), "\n"))
  if (p_sector < alpha) {
    cat("  El Efecto Principal del Sector es SIGNIFICATIVO (p < ", alpha, ").\n")
    cat("  Existen diferencias significativas en las notas de matemática entre el sector público y privado.\n")
    # Para 2 grupos, no se necesitan post-hoc adicionales, la tabla ANOVA ya lo dice.
    # Pero si quieres las medias para la interpretación:
    cat("\nMedias de Notas de Matemática por Sector:\n")
    print(aggregate(Promedio ~ EsPrivada, data = datos_anova, FUN = mean))
  } else {
    cat("  El Efecto Principal del Sector NO es significativo (p >= ", alpha, ").\n")
    cat("  No hay evidencia suficiente para concluir que existen diferencias significativas en las notas de matemática entre los sectores.\n")
  }
}
```

```r
install_and_load <- function(package_name) {
  if (!require(package_name, character.only = TRUE)) {
    # Si el paquete no está cargado, intenta instalarlo
    message(paste("Instalando el paquete:", package_name))
    install.packages(package_name, dependencies = TRUE) # dependencies = TRUE para instalar dependencias también

    # Después de instalar, intenta cargarlo de nuevo
    if (!require(package_name, character.only = TRUE)) {
      stop(paste("No se pudo instalar ni cargar el paquete:", package_name, ". Por favor, inténtelo manualmente."))
    }
  }
}

# --- 3. Comparar los Vectores de Medias y Test de Hipótesis (MANOVA) ---

cat("\n--- Análisis de MANOVA para Comparar Vectores de Medias ---\n")

# 3.1. Verificar Suposiciones del MANOVA

# a) Normalidad Univariada (por grupo)
cat("\nVerificación de Normalidad (univariada, por grupo - Prueba de Shapiro-Wilk):\n")

shapiro_results <- dataset_1ro_secundaria %>%
  group_by(EsPrivada) %>%
  summarise(
    # Para cada materia, aplica shapiro.test() y extrae directamente el p-valor.
    # El '.names' asegura nombres de columna únicos para cada materia (ej., Shapiro_p_PromedioMatematica)
    across(all_of(materias_troncales), ~ shapiro.test(.x)$p.value, .names = "Shapiro_p_{.col}")
  )

print(shapiro_results)

cat("  - Si el p-valor de Shapiro-Wilk (las columnas que empiezan con 'Shapiro_p_') es < 0.05 para alguna materia en algún grupo, sugiere no normalidad univariada.\n")
cat("  - Para MANOVA, buscamos normalidad multivariada, que es más estricta. Sin embargo, MANOVA es algo robusto a pequeñas desviaciones de la normalidad univariada para muestras grandes (>20-30 observaciones por grupo).\n")
cat("  - Alternativamente, puedes visualizar con gráficos Q-Q para una inspección visual más detallada si hay preocupaciones.\n")

# b) Homogeneidad de Matrices de Covarianza (Test M de Box)
# H0: Las matrices de covarianza son iguales entre los grupos.
# H1: Al menos una matriz de covarianza difiere.
cat("\nTest M de Box para Homogeneidad de Matrices de Covarianza:\n")
# Usamos la función BoxMTest del paquete heplots
# x: las variables dependientes
# group: la variable de agrupación
# Test de noralidad de marbia
box_m_test <- BoxMTest(dataset_1ro_secundaria[, materias_troncales], cl = as.factor(dataset_1ro_secundaria$EsPrivada))
print(box_m_test)

#accedemos al p value del box test.
p_valor_box_m <- box_m_test$pValue # Asegúrate que el nombre sea "p.value" o "p-value"

if (p_valor_box_m < 0.001) { # Box's M suele evaluarse con un alpha más estricto, ej. 0.001
  cat("  ¡Advertencia! El p-valor del Test M de Box es < 0.001 (significativo). Se RECHAZA H0.\n")
  cat("  Esto sugiere que las matrices de covarianza NO son homogéneas. Esto es una violación de la suposición importante del MANOVA clásico.\n")
  cat("  Considerar usar el estadístico de Pillai's Trace en el MANOVA, ya que es más robusto a esta violación, o métodos robustos de MANOVA.\n")
} else {
  cat("  El p-valor del Test M de Box es >= 0.001 (no significativo). No hay evidencia para rechazar H0.\n")
  cat("  Se asume la homogeneidad de las matrices de covarianza. El MANOVA clásico es apropiado.\n")
}

# 3.2. Realizar el MANOVA
cat("\n--- Resultados del MANOVA ---\n")
# La sintaxis es: manova(cbind(dep_var1, dep_var2, ...) ~ indep_var, data = dataframe)
manova_result <- manova(as.matrix(dataset_1ro_secundaria[, materias_troncales]) ~ EsPrivada, data = dataset_1ro_secundaria)

# Imprimir el resumen del MANOVA
# summary(manova_result) mostrará Wilks' Lambda por defecto
print(summary(manova_result))

# También puedes especificar el estadístico de prueba. Pillai's Trace es más robusto.
cat("\nResumen del MANOVA (usando Pillai's Trace - más robusto):\n")
print(summary(manova_result, test = "Pillai"))

# --- 3.3. Interpretación del MANOVA Global (Consigna b) ---
cat("\n--- Interpretación del Test de Hipótesis (Consigna b) ---\n")

# Obtener el p-valor de Wilks' Lambda (o el test que elijas)
manova_summary_wilks <- summary(manova_result)$stats
p_value_manova <- manova_summary_wilks["EsPrivada", "Pr(>F)"] # Obtener el p-valor de Wilks' Lambda

cat(paste0("  Valor p del MANOVA (Wilks' Lambda): ", round(p_value_manova, 4), "\n"))
alpha_manova <- 0.05 # Nivel de significancia para el MANOVA
cat(paste0("  Nivel de significancia (alpha) para el MANOVA: ", alpha_manova, "\n"))

if (p_value_manova < alpha_manova) {
  cat("  Se RECHAZA la hipótesis nula (H0). Conclusión:\n")
  cat("  Existen diferencias SIGNIFICATIVAS en los vectores de medias de las notas de materias troncales entre los sectores (público vs. privado).\n")
  cat("  Esto significa que el perfil global de notas difiere entre las escuelas públicas y privadas.\n")

  # --- Si el MANOVA es significativo, realizar ANOVAs univariados de seguimiento ---
  cat("\n--- ANOVAs Univariados de Seguimiento (si MANOVA es significativo) ---\n")
  cat("  Para identificar qué materias específicas contribuyen a la diferencia global:\n")

  # Realizar ANOVAs univariados para cada materia
  univariate_results <- data.frame(
    Materia = character(),
    F_value = numeric(),
    p_value = numeric(),
    p_adj_bonferroni = numeric()
  )

  for (materia in materias_troncales) {
    formula <- as.formula(paste(materia, "~ EsPrivada"))
    aov_res <- aov(formula, data = dataset_1ro_secundaria)
    summary_aov <- summary(aov_res)
    f_val <- summary_aov[[1]]$`F value`[1]
    p_val <- summary_aov[[1]]$`Pr(>F)`[1]

    univariate_results <- rbind(univariate_results, data.frame(
      Materia = materia,
      F_value = f_val,
      p_value = p_val,
      p_adj_bonferroni = NA # Se calculará después
    ))
  }

  # Ajustar p-valores para comparaciones múltiples (Bonferroni)
  univariate_results$p_adj_bonferroni <- p.adjust(univariate_results$p_value, method = "bonferroni")

  cat("\nResultados de ANOVAs Univariados (con corrección de Bonferroni):\n")
  print(univariate_results)

  cat("\nInterpretación de ANOVAs Univariados:\n")
  cat("  - Las materias con 'p_adj_bonferroni' menor que ", alpha_manova, " muestran diferencias significativas entre los sectores.\n")
  cat("  - Por ejemplo, si 'PromedioMatematica' tiene un p_adj_bonferroni < 0.05, significa que las notas promedio de Matemáticas difieren significativamente entre escuelas públicas y privadas.\n")

  # Para ver las medias de cada grupo para las materias significativas
  cat("\nMedias por Sector para cada Materia:\n")
  print(dataset_1ro_secundaria %>% group_by(EsPrivada) %>% summarise(across(all_of(materias_troncales), mean, .names = "Media_{.col}")))

} else {
  cat("  NO se RECHAZA la hipótesis nula (H0). Conclusión:\n")
  cat("  No existen diferencias SIGNIFICATIVAS en los vectores de medias de las notas de materias troncales entre los sectores (público vs. privado) en el conjunto de variables.\n")
  cat("  Esto significa que el perfil global de notas no difiere significativamente entre las escuelas públicas y privadas.\n")
}
```

```r
# --- a) Indique su vector de medias y medianas ---
cat("\n--- Vectores de Medias y Medianas por Año de Cursado ---\n")

summary_notas_por_anio <- dataset_secundaria_completa %>%
  group_by(Anio) %>%
  summarise(
    across(all_of(materias_troncales), .fns = list(
      Media = mean,
      Mediana = median
    ), .names = "{.col}_{.fn}")
  ) %>%
  ungroup()

# Mostrar la tabla de forma legible
cat("\nMedias y Medianas por Año:\n")
print(kable(summary_notas_por_anio, format = "pipe", digits = 2))


# --- b) Calcule su matriz de varianza y covarianza ---
cat("\n\n--- Matriz de Varianza y Covarianza ---\n")

# Para el conjunto completo de datos (sin agrupar por año)
cat("\nMatriz de Varianza-Covarianza General (sin agrupar por Año):\n")
matriz_cov_general <- cov(dataset_secundaria_completa[, materias_troncales])
print(round(matriz_cov_general, 2))
cat("  - La diagonal principal muestra las varianzas de cada materia.\n")
cat("  - Los valores fuera de la diagonal muestran las covarianzas entre pares de materias.\n")


# Matriz de Varianza-Covarianza Pooled (Agrupada por Año - CÁLCULO MANUAL CORREGIDO)
# Esta matriz representa una estimación combinada de la covarianza entre las materias,
# asumiendo (o aproximando) que la estructura de covarianza es similar en todos los años de cursado.
cat("\nMatriz de Varianza-Covarianza Pooled (Agrupada por Año - Cálculo Manual):\n")

# Obtener los años únicos para agrupar
unique_years <- levels(dataset_secundaria_completa$Anio)
if (is.null(unique_years)) { # En caso de que Anio no sea un factor o no tenga niveles
  unique_years <- unique(dataset_secundaria_completa$Anio)
}

# Inicializar una matriz de ceros para sumar las covarianzas ponderadas
pooled_cov_sum <- matrix(0, nrow = length(materias_troncales), ncol = length(materias_troncales))
colnames(pooled_cov_sum) <- materias_troncales
rownames(pooled_cov_sum) <- materias_troncales

total_df <- 0 # Variable para sumar los grados de libertad de cada grupo

# Iterar sobre cada año para calcular la covarianza y sumarla
for (year_level in unique_years) {
  group_data <- dataset_secundaria_completa %>% filter(Anio == year_level)
  n_k <- nrow(group_data) # Número de observaciones en el grupo actual

  # Solo calcula la covarianza si hay suficientes datos (al menos 2 observaciones)
  if (n_k > 1) {
    cov_k <- cov(group_data[, materias_troncales]) # Matriz de covarianza para el grupo actual
    pooled_cov_sum <- pooled_cov_sum + (n_k - 1) * cov_k # Sumar ponderando por grados de libertad
    total_df <- total_df + (n_k - 1) # Sumar grados de libertad
  } else {
    # Mensaje si un grupo es demasiado pequeño
    cat(paste("  Advertencia: El año", year_level, "tiene solo", n_k, "observación(es). No se puede calcular su covarianza.\n"))
  }
}

# Calcular la matriz pooled final si tenemos grados de libertad suficientes
if (total_df > 0) {
  pooled_cov_matrix_manual <- pooled_cov_sum / total_df
  print(round(pooled_cov_matrix_manual, 2))
  cat("  - Esta matriz es el promedio ponderado de las matrices de covarianza de cada año.\n")
  cat("  - Se utiliza en MANOVA para la prueba de hipótesis si se asume homogeneidad de varianzas.\n")
} else {
  cat("\nNo se pudo calcular la Matriz de Varianza-Covarianza Pooled. No hay suficientes datos en los grupos para ello.\n")
}
```

### PARTE 2 --> Realice un test que permita contrastar si existen diferencias significativas entre los promedios de los diferentes años de cursado en las materias troncales de Secundaria.


```r
# --- 0. Cargar librerías necesarias ---
# Asegúrate de que estas librerías estén instaladas y cargadas
library(dplyr)
library(heplots) # Para BoxMTest
library(rstatix) # Para shapiro_test y otras utilidades si se necesitan


# --- 1. Definir las variables dependientes ---
materias_troncales <- c("PromedioMatematica", "PromedioLengua", "PromedioGeografia", "PromedioHistoria")

# --- 2. Verificación de Suposiciones del MANOVA ---

cat("\n--- Verificación de Suposiciones para MANOVA ---\n")

# a) Normalidad Univariada (por grupo - Prueba de Shapiro-Wilk)
cat("\nPrueba de Normalidad Univariada (Shapiro-Wilk) por Año:\n")
# Se aplica el test de Shapiro-Wilk a cada materia dentro de cada año
shapiro_results_anio <- dataset_secundaria_completa %>%
  group_by(Anio) %>%
  summarise(
    across(all_of(materias_troncales), ~ shapiro.test(.x)$p.value, .names = "Shapiro_p_{.col}")
  )

print(shapiro_results_anio)

cat("  - Si el p-valor de Shapiro-Wilk (las columnas que empiezan con 'Shapiro_p_') es < 0.05 para alguna materia en algún año, sugiere no normalidad univariada en ese grupo.\n")
cat("  - MANOVA asume normalidad multivariada, que es más estricta. Sin embargo, MANOVA es algo robusto a pequeñas desviaciones de la normalidad univariada para tamaños de muestra grandes (>20-30 observaciones por grupo).\n")
cat("  - Visualiza con gráficos Q-Q si tienes preocupaciones serias sobre la normalidad.\n")


# b) Homogeneidad de Matrices de Covarianza (Test M de Box)
cat("\nTest M de Box para Homogeneidad de Matrices de Covarianza por Año:\n")
# Asegúrate de que library(heplots) se haya ejecutado correctamente
box_m_test_anio <- BoxMTest(dataset_secundaria_completa[, materias_troncales], cl = as.factor(dataset_secundaria_completa$Anio))
print(box_m_test_anio)

# Interpretar el resultado del Test M de Box
alpha_box_m <- 0.001 # Se usa un alpha más estricto para Box's M debido a su sensibilidad
if (box_m_test_anio$pValue < alpha_box_m) {
  cat(paste0("  ¡Advertencia! El p-valor del Test M de Box es < ", alpha_box_m, " (p = ", round(box_m_test_anio$pValue, 5), "). Se RECHAZA H0.\n"))
  cat("  Esto sugiere que las matrices de covarianza NO son homogéneas entre los años. Esta es una violación de la suposición del MANOVA clásico.\n")
  cat("  Por lo tanto, se recomienda interpretar el estadístico de prueba **Pillai's Trace** en el MANOVA, ya que es más robusto a esta violación.\n")
} else {
  cat(paste0("  El p-valor del Test M de Box es >= ", alpha_box_m, " (p = ", round(box_m_test_anio$pValue, 5), "). No hay evidencia para rechazar H0.\n"))
  cat("  Se asume la homogeneidad de las matrices de covarianza. El MANOVA clásico con Wilks' Lambda es apropiado.\n")
}
```

```r
# --- 3. Realizar el MANOVA ---
cat("\n--- Resultados del Análisis Multivariado de Varianza (MANOVA) ---\n")

# La fórmula indica que las materias troncales (combinadas con cbind) son las VD
# y Anio es la VI (variable independiente de agrupación).
manova_result_anio <- manova(as.matrix(dataset_secundaria_completa[, materias_troncales]) ~ Anio, data = dataset_secundaria_completa)

# Mostrar el resumen del MANOVA con los estadísticos predeterminados (Wilks' Lambda es común)
cat("\nResumen del MANOVA (Estadístico por defecto, usualmente Wilks' Lambda):\n")
print(summary(manova_result_anio))

# Mostrar el resumen del MANOVA usando Pillai's Trace (más robusto si Box's M es significativo)
cat("\nResumen del MANOVA (usando Pillai's Trace - recomendado si Box's M es significativo):\n")
print(summary(manova_result_anio, test = "Pillai"))
```

```r
# --- 4. Interpretación de los Resultados ---

cat("\n--- Interpretación de los Resultados del MANOVA ---\n")

# Obtener el p-valor del estadístico de interés (usaremos Pillai's Trace por robustez)
manova_summary_pillai <- summary(manova_result_anio, test = "Pillai")$stats
p_value_manova_pillai <- manova_summary_pillai["Anio", "Pr(>F)"]

alpha_manova <- 0.05
cat(paste0("Nivel de significancia (alpha) para el MANOVA: ", alpha_manova, "\n"))
cat(paste0("P-valor (Pillai's Trace): ", round(p_value_manova_pillai, 4), "\n"))

if (p_value_manova_pillai < alpha_manova) {
  cat("\nConclusión del MANOVA Global (Pillai's Trace):\n")
  cat("Se RECHAZA la hipótesis nula ($H_0$).\n")
  cat(paste0("  Existe(n) diferencia(s) SIGNIFICATIVA(s) en el vector de medias de las notas de las materias troncales entre los diferentes años de cursado de secundaria (p < ", alpha_manova, ").\n"))
  cat("  Esto significa que el perfil de notas promedio de los estudiantes en las materias troncales difiere significativamente en al menos un par de años de secundaria.\n")

  cat("\n--- Análisis de Seguimiento (si el MANOVA global es significativo) ---\n")
  cat("  Para identificar qué materias específicas contribuyen a la diferencia global y entre qué años:\n")

  # a) ANOVAs Univariados para cada materia (con corrección de Bonferroni)
  # Esto nos dice si cada materia, por sí sola, muestra diferencias significativas entre los años.
  cat("\nResultados de ANOVAs Univariados (con corrección de Bonferroni para el p-valor):\n")

  univariate_aov_results <- data.frame(
    Materia = character(),
    F_value = numeric(),
    p_value_unadjusted = numeric(),
    p_value_adjusted = numeric()
  )

  for (materia in materias_troncales) {
    formula_aov <- as.formula(paste(materia, "~ Anio"))
    aov_res <- aov(formula_aov, data = dataset_secundaria_completa)
    summary_aov <- summary(aov_res)
    f_val <- summary_aov[[1]]$`F value`[1]
    p_val <- summary_aov[[1]]$`Pr(>F)`[1]

    univariate_aov_results <- rbind(univariate_aov_results, data.frame(
      Materia = materia,
      F_value = f_val,
      p_value_unadjusted = p_val,
      p_value_adjusted = NA # Se ajustará después
    ))
  }
  # Aplicar corrección de Bonferroni (para controlar el error Tipo I al hacer múltiples tests)
  univariate_aov_results$p_value_adjusted <- p.adjust(univariate_aov_results$p_value_unadjusted, method = "bonferroni")

  print(univariate_aov_results)
  cat(paste0("  - Las materias con 'p_value_adjusted' menor que ", alpha_manova, " indican diferencias significativas en el promedio de esa materia entre los años.\n"))


  # b) Pruebas Post-Hoc (ej. Tukey HSD) para las materias que resultaron significativas en el ANOVA univariado
  # Esto nos dice entre qué años específicos existen las diferencias para cada materia significativa.
  cat("\nPruebas Post-Hoc (Tukey HSD) para materias con ANOVA univariado significativo:\n")

  for (i in 1:nrow(univariate_aov_results)) {
    if (univariate_aov_results$p_value_adjusted[i] < alpha_manova) {
      materia_sig <- univariate_aov_results$Materia[i]
      cat(paste0("\n--- Post-Hoc para ", materia_sig, " ---\n"))
      formula_tukey <- as.formula(paste(materia_sig, "~ Anio"))
      tukey_hsd_res <- TukeyHSD(aov(formula_tukey, data = dataset_secundaria_completa))
      print(tukey_hsd_res$Anio)
      cat("  - En la tabla de Tukey HSD, busca las filas donde 'p adj' (p-valor ajustado) sea menor que ", alpha_manova, ".\n")
      cat("  - Esto te indicará qué pares de años tienen diferencias significativas en el promedio de ", materia_sig, ".\n")
    }
  }

} else {
  cat("\nConclusión del MANOVA Global (Pillai's Trace):\n")
  cat("NO se RECHAZA la hipótesis nula ($H_0$).\n")
  cat(paste0("  No existe evidencia significativa de diferencias en el vector de medias de las notas de las materias troncales entre los diferentes años de cursado de secundaria (p >= ", alpha_manova, ").\n"))
  cat("  Esto significa que el perfil de notas promedio de los estudiantes en las materias troncales no difiere significativamente en el conjunto de años.\n")
}
```

```r
# Verificar e instalar 'scatterplot3d' si no está presente
if (!requireNamespace("scatterplot3d", quietly = TRUE)) {
  install.packages("scatterplot3d")
}
# Ya se asume 'dplyr' y 'aplpack' están instalados y cargados de los pasos anteriores

library(scatterplot3d)
library(dplyr) # Asegurarse de que dplyr esté cargado

# --- Paso 1: Obtener el vector de medias de primer año de cada escuela ---
materias_troncales <- c('PromedioMatematica', 'PromedioLengua', 'PromedioGeografia', 'PromedioHistoria')

data_agrupada_por_escuela <- dataset_1ro_secundaria %>%
  group_by(suborganizacion) %>%
  summarise(
    PromedioMatematica = mean(PromedioMatematica, na.rm = TRUE),
    PromedioLengua = mean(PromedioLengua, na.rm = TRUE),
    PromedioGeografia = mean(PromedioGeografia, na.rm = TRUE),
    PromedioHistoria = mean(PromedioHistoria, na.rm = TRUE),
    # ¡¡¡CORRECCIÓN CLAVE AQUÍ: Eliminado na.rm = TRUE de first()!!!
    # Se toma el primer valor de EsPrivada para cada escuela.
    EsPrivada = first(EsPrivada)
  ) %>%
  ungroup()

# --- Paso 2: Preparar los datos para el gráfico 3D ---
data_3d <- data_agrupada_por_escuela %>%
  select(PromedioMatematica, PromedioLengua, PromedioHistoria)

nombres_escuelas_unicos_cortos <- make.unique(substr(data_agrupada_por_escuela$suborganizacion, 1, 20))
rownames(data_3d) <- nombres_escuelas_unicos_cortos

# --- Paso 3: Realizar el Gráfico 3D ---

colores_escuelas <- data_agrupada_por_escuela$EsPrivada

# Convertir 'EsPrivada' a colores
# Asegúrate de que los valores en tu columna 'EsPrivada' son exactamente 'N' y 'S'.
# Si tienes NA en EsPrivada, esos puntos se asignarán a "red" (Privada) con esta lógica.
# Si prefieres que NA sea un color diferente, puedes ajustar la condición.
colores_plot <- ifelse(colores_escuelas == 'N', 'blue', 'red')

# Definir la forma de los puntos (pch)
puntos_plot <- ifelse(colores_escuelas == 'N', 16, 15)

# Crear el gráfico 3D
s3d <- scatterplot3d(x = data_3d$PromedioMatematica,
                     y = data_3d$PromedioLengua,
                     z = data_3d$PromedioHistoria,
                     xlab = "Promedio Matemática",
                     ylab = "Promedio Lengua",
                     zlab = "Promedio Historia",
                     main = "Gráfico 3D de Promedios de Materias por Escuela",
                     col.grid = "lightblue",
                     color = colores_plot,
                     pch = puntos_plot,
                     type = "h",
                     highlight.3d = TRUE,
                     box = TRUE,
                     grid = TRUE,
                     angle = 50,
                     cex.symbols = 1.2
                    )

# Añadir una leyenda para los colores
legend("topright", legend = c("No Privada", "Privada"),
       col = c("blue", "red"), pch = c(16, 15), title = "Tipo de Escuela")

# --- Interpretación del gráfico 3D ---
cat("\n\n--- Interpretación del Gráfico 3D ---\n")
cat("Este gráfico 3D visualiza los promedios de tres materias (Matemáticas, Lengua e Historia) para cada escuela.\n")
cat("Cada punto representa una escuela. Su posición en el espacio 3D indica sus promedios en estas tres materias.\n")
cat(" - El eje X (horizontal): Promedio de Matemática.\n")
cat(" - El eje Y (profundidad): Promedio de Lengua.\n")
cat(" - El eje Z (vertical): Promedio de Historia.\n")
cat("\nLos puntos están coloreados para diferenciar entre escuelas No Privadas (azul) y Privadas (rojo).\n")
cat("Además, las escuelas No Privadas son círculos y las Privadas son cuadrados.\n")
cat("\n**¿Cómo visualizar las diferencias?**\n")
cat(" - **Escuelas similares:** Los puntos que están muy cerca unos de otros en el espacio 3D representan escuelas con perfiles de notas promedio muy similares en las tres materias seleccionadas.\n")
cat(" - **Escuelas con altos promedios:** Buscar puntos que estén en la parte superior derecha y hacia adelante del gráfico (altos valores en los tres ejes).\n")
cat(" - **Escuelas con bajos promedios:** Buscar puntos que estén en la parte inferior izquierda y hacia atrás del gráfico (bajos valores en los tres ejes).\n")
cat(" - **Diferencias notorias:** Observa si hay grupos de puntos que se aglomeran en diferentes regiones del espacio. Por ejemplo:\n")
cat("    - Un grupo de puntos en la esquina superior que son predominantemente rojos (privadas con altos promedios).\n")
cat("    - Otro grupo de puntos en la parte inferior que son predominantemente azules (no privadas con bajos promedios).\n")
cat(" - **Relaciones entre materias:** La orientación del gráfico permite ver cómo se relacionan los promedios. Por ejemplo, si los puntos tienden a seguir una diagonal, podría indicar que las escuelas con alto promedio en una materia también lo tienen en las otras.\n")
cat(" - **Diferencias por tipo de escuela:** Observa si los colores (azul/rojo) o las formas (círculo/cuadrado) se agrupan en distintas regiones del gráfico, lo que indicaría diferencias en los perfiles de notas entre escuelas privadas y no privadas.\n")
```
